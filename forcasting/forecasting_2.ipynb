```python
from pyspark.sql.types import *

toner_usage_schema = StructType([

    StructField("deviceId", StringType(), True),

    StructField("dealerId", LongType(), True),
    StructField("L1DealerId", LongType(), True),
    StructField("L2DealerId", LongType(), True),
    StructField("L3DealerId", LongType(), True),

    StructField("divisionId", StringType(), True),
    StructField("servicingDealerId", StringType(), True),
    StructField("customerId", LongType(), True),

    StructField("modelName", StringType(), True),
    StructField("serialNumber", LongType(), True),

    # --- Forecasting related fields ---
    StructField("timestamp", TimestampType(), True),
    StructField("lastSuppliesUpdate", TimestampType(), True),

    StructField("relatedGroupId", StringType(), True),
    StructField("type", IntegerType(), True),

    StructField("description", StringType(), True),
    StructField("state", StringType(), True),
    StructField("color", StringType(), True),

    StructField("typical", IntegerType(), True),   # Toner %
    StructField("unit", StringType(), True),
    StructField("capacity", IntegerType(), True),

    StructField("printCount", LongType(), True),
    StructField("tonerNumber", IntegerType(), True),

    StructField("tagIds", StringType(), True),
    StructField("forecastingList", StringType(), True)
])
```

```python
df_toner = spark.read \
    .format("csv") \
    .option("header", "true") \
    .schema(toner_usage_schema) \
    .load("/Volumes/workspace/default/csv/Toner Usage.csv")
```

```python
df_toner.printSchema()

display(df_toner)
```

```python
from pyspark.sql.functions import col, to_timestamp

df_selected = df_toner.select(
    "deviceId",
    "color",
    "timestamp",
    "typical",
    "printCount"
)
display(df_selected)
```

```python
from pyspark.sql.functions import unix_timestamp

df_prepared = (
    df_selected
    .withColumn("timestamp", to_timestamp("timestamp"))
    .withColumn("time_sec", unix_timestamp("timestamp"))
    .withColumn("typical", col("typical").cast("double"))
    .withColumn("printCount", col("printCount").cast("double"))
)

```

```python
df_clean = (
    df_prepared
    .filter(col("typical").between(0, 100))
    .filter(col("timestamp").isNotNull())
    .filter(col("printCount").isNotNull())
)
display(df_clean)
```

```python
df_dedup = df_clean.dropDuplicates(
    ["deviceId", "color", "timestamp", "typical"]
)
display(df_dedup)
```

```python
df_sorted = df_dedup.orderBy(
    "deviceId", "color", "timestamp"
)
```

```python
data = df_sorted.select(
    "time_sec", "typical", "printCount"
).collect()

x_time = [row.time_sec for row in data]
y_typical = [row.typical for row in data]
y_print = [row.printCount for row in data]
display(df_sorted.select("time_sec", "typical", "printCount"))
```

```python
def simple_linear_regression(x, y):
    n = len(x)
    sum_x = sum(x)
    sum_y = sum(y)
    sum_xy = sum(x[i] * y[i] for i in range(n))
    sum_x2 = sum(x[i] ** 2 for i in range(n))

    m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)
    c = (sum_y - m * sum_x) / n

    return m, c
```

```python
m_time, c_time = simple_linear_regression(x_time, y_typical)

print("Slope (m):", m_time)
print("Intercept (c):", c_time)
```

```python
m_time, c_time = simple_linear_regression(x_time, y_typical)

predicted_time_sec = -c_time / m_time
```

```python
from datetime import datetime

predicted_end_date = datetime.fromtimestamp(predicted_time_sec)
print("Predicted Toner End Date:", predicted_end_date)
```

```python
m_print, c_print = simple_linear_regression(y_typical, y_print)

predicted_print_count = c_print
print("Predicted Print Count at Toner End:", predicted_print_count)
```

```python
devices = (
    df_sorted
    .select("deviceId", "color")
    .distinct()
    .collect()
)
```

```python
from datetime import datetime

results = []

for row in devices:
    device_id = row.deviceId
    color = row.color

    # Filter data for one device + color
    device_df = df_sorted.filter(
        (col("deviceId") == device_id) &
        (col("color") == color)
    )

    # Collect data
    data = device_df.select("time_sec", "typical").collect()

    # Need at least 2 points for regression
    if len(data) < 2:
        continue

    x_time = [r.time_sec for r in data]
    y_typical = [r.typical for r in data]

    # Apply simple linear regression
    m_time, c_time = simple_linear_regression(x_time, y_typical)

    # Skip invalid slope
    if m_time == 0:
        continue

    # Predict time when Typical = 0
    predicted_time_sec = -c_time / m_time
    predicted_end_date = datetime.fromtimestamp(predicted_time_sec)

    results.append(
        (device_id, color, predicted_end_date)
    )
```

```python
forecast_df = spark.createDataFrame(
    results,
    ["deviceId", "color", "predicted_end_date"]
)

display(forecast_df)
```
