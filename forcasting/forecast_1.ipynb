```python
from pyspark.sql.types import *

toner_usage_schema = StructType([

    StructField("deviceId", StringType(), True),

    StructField("dealerId", LongType(), True),
    StructField("L1DealerId", LongType(), True),
    StructField("L2DealerId", LongType(), True),
    StructField("L3DealerId", LongType(), True),

    StructField("divisionId", StringType(), True),
    StructField("servicingDealerId", StringType(), True),
    StructField("customerId", LongType(), True),

    StructField("modelName", StringType(), True),
    StructField("serialNumber", LongType(), True),

    # --- Forecasting related fields ---
    StructField("timestamp", TimestampType(), True),
    StructField("lastSuppliesUpdate", TimestampType(), True),

    StructField("relatedGroupId", StringType(), True),
    StructField("type", IntegerType(), True),

    StructField("description", StringType(), True),
    StructField("state", StringType(), True),
    StructField("color", StringType(), True),

    StructField("typical", IntegerType(), True),   # Toner %
    StructField("unit", StringType(), True),
    StructField("capacity", IntegerType(), True),

    StructField("printCount", LongType(), True),
    StructField("tonerNumber", IntegerType(), True),

    StructField("tagIds", StringType(), True),
    StructField("forecastingList", StringType(), True)
])
```

```python
df_toner = spark.read \
    .format("csv") \
    .option("header", "true") \
    .schema(toner_usage_schema) \
    .load("/Volumes/workspace/default/csv/Toner Usage.csv")
```

```python
df_toner.printSchema()

display(df_toner)
```

```python
from pyspark.sql.functions import to_timestamp, unix_timestamp, col

df_prepared = (
    df_toner.select("deviceId", "color", "timestamp", "typical", "printCount")
    .withColumn("timestamp", to_timestamp("timestamp"))
    .withColumn("time_sec", unix_timestamp("timestamp"))
    .withColumn("typical", col("typical").cast("double"))
    .withColumn("printCount", col("printCount").cast("double"))
    .filter(col("typical").between(0, 100))
)
```

```python
df_sorted = (
    df_prepared
    .dropDuplicates(["deviceId", "color", "timestamp", "typical"])
    .orderBy("deviceId", "color", "timestamp")
)
```

```python
from pyspark.sql.functions import col

# Get unique deviceId and color combinations
df_devices = (
    df_sorted
    .select("deviceId", "color")
    .distinct()
)

devices = df_devices.collect()

math_results = []

for row in devices:
    device_id = row.deviceId
    color = row.color

    device_df = (
        df_sorted
        .filter(
            (col("deviceId") == device_id) &
            (col("color") == color)
        )
        .orderBy("timestamp")
    )

    data = device_df.select(
        "time_sec", "typical", "printCount"
    ).collect()

    predicted_end_date = None
    predicted_print_count = None

    if len(data) >= 2:
        total_toner_drop = 0
        total_time = 0
        total_print = 0

        for i in range(1, len(data)):
            t_diff = data[i].time_sec - data[i-1].time_sec
            toner_diff = data[i-1].typical - data[i].typical
            print_diff = data[i].printCount - data[i-1].printCount

            if t_diff > 0 and toner_diff > 0:
                total_toner_drop += toner_diff
                total_time += t_diff
                total_print += print_diff

        if total_time > 0 and total_toner_drop > 0:
            avg_toner_rate = total_toner_drop / total_time
            remaining_time_sec = data[-1].typical / avg_toner_rate

            predicted_time_sec = data[-1].time_sec + remaining_time_sec
            predicted_end_date = datetime.fromtimestamp(predicted_time_sec)

            avg_print_rate = total_print / total_time
            predicted_print_count = (
                data[-1].printCount + avg_print_rate * remaining_time_sec
            )

    math_results.append(
        (
            device_id,
            color,
            predicted_end_date,
            predicted_print_count
        )
    )
```

```python
math_forecast_df = spark.createDataFrame(
    math_results,
    [
        "deviceId",
        "color",
        "predicted_end_date",
        "predicted_print_count"
    ]
)

display(math_forecast_df)
```
