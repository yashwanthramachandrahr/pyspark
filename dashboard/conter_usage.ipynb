```python
from pyspark.sql.functions import col, to_timestamp, to_date, month, year

json_path = "/Volumes/workspace/default/json/counterUsage.json"

raw_df = spark.read \
    .option("multiline", "true") \
    .json(json_path)

display(raw_df)
```

```python
from pyspark.sql.functions import lit

transformed_df = raw_df.select(
    col("relatedAgentId").alias("customer_id"),
    col("friendlyName").alias("customer_name"),
    col("deviceId").alias("device_id"),
    col("modelName").alias("model_name"),
    col("friendlyName"),

    to_timestamp(col("timestamp.$date")).alias("event_timestamp"),

    col("dailyUsage.total.output.value").alias("output_total"),
    col("dailyUsage.total.send.value").alias("send_total")
)
display(transformed_df)
```

```python
final_df = transformed_df \
    .withColumn("event_date", to_date("event_timestamp")) \
    .withColumn("event_month", month("event_timestamp")) \
    .withColumn("event_year", year("event_timestamp")) \
    .withColumn("total_usage", col("output_total") + col("send_total")) \
    .fillna(0)

display(final_df)
```

```python
%sql
CREATE DATABASE IF NOT EXISTS dashboard_db;

```

```python
final_df.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable("dashboard_db.device_usage")
```
